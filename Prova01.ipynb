#Qual estrutura é mais adequada para guardar os chamados já atendidos, considerando que o sistema precisa desfazer primeiro o atendimento mais recente? Justifique
R: A estrutura mais adequada é a pilha (Stack), porque precisamos desfazer o último atendimento realizado primeiro. Isso segue a regra LIFO, “último a entrar, primeiro a sair”.
#Sem executar código, indique como estarão as duas estruturas após a sequencia a seguir : atender atender desfazer atender
R :No começo, a fila possui os chamados 101, 102, 103 e 104.

Primeiro atender: Ana (101) é atendida.

Segundo atender: Bruno (102) é atendido.

Desfazer: o último atendimento, Bruno (102), é desfeito e ele volta para o início da fila.

Terceiro atender: Bruno (102) é atendido novamente.
Fila de espera: 103 - Carla; 104 - Daniel.
Pilha de atendidos: 101 - Ana; 102 - Bruno.


```python
#parte 2
class CentralAtendimento:

    def __init__(self):
        self.fila = []
        self.atendidos = []

    def adicionar_chamado(self, codigo, cliente, prioridade):
        chamado = {
            "codigo": codigo,
            "cliente": cliente,
            "prioridade": prioridade
        }
        self.fila.append(chamado)

    def atender_proximo(self):
        if len(self.fila) > 0:
            chamado = self.fila.pop(0)
            self.atendidos.append(chamado)
            return chamado
        else:
            print("Não há chamados na fila.")

    def desfazer_ultimo_atendimento(self):
        if len(self.atendidos) > 0:
            chamado = self.atendidos.pop()
            self.fila.insert(0, chamado)
            return chamado
        else:
            print("Não há atendimentos para desfazer.")

    def buscar_chamado(self, codigo):
        for chamado in self.fila:
            if chamado["codigo"] == codigo:
                return chamado

        for chamado in self.atendidos:
            if chamado["codigo"] == codigo:
                return chamado

        return "Chamado não encontrado."

    def exibir_situacao(self):
        print("Fila de espera:")
        for chamado in self.fila:
            print(chamado)

        print("\nChamados atendidos:")
        for chamado in self.atendidos:
            print(chamado)
```

```python
#parte 3
central = CentralAtendimento()

central.adicionar_chamado(101, "Ana", "Média")
central.adicionar_chamado(102, "Bruno", "Alta")
central.adicionar_chamado(103, "Carla", "Baixa")
central.adicionar_chamado(104, "Daniel", "Alta")

central.atender_proximo()
central.atender_proximo()

central.desfazer_ultimo_atendimento()

central.atender_proximo()

central.exibir_situacao()

print(central.buscar_chamado(104))
```
